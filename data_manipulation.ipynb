```python
import os
import csv
import re

def parse_files_to_csv(folder_path, output_csv):
    # Regex to match: id_start_end (no extension required)
    pattern = re.compile(r"^([^_]+)_([^_]+)_([^_]+)$")

    rows = []

    for filename in os.listdir(folder_path):
        name, ext = os.path.splitext(filename)
        match = pattern.match(name)

        if match:
            file_id, start, end = match.groups()
            rows.append([file_id, start, end, filename])

    # Write CSV
    with open(output_csv, "w", newline="", encoding="utf-8") as f:
        writer = csv.writer(f)
        writer.writerow(["id", "start", "end", "filename"])
        writer.writerows(rows)

    print(f"CSV created: {output_csv}")


# Example usage
if __name__ == "__main__":
    parse_files_to_csv(
        folder_path="data/part_241",
        output_csv="output.csv"
    )
```
